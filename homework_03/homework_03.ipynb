## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
from homework_02.h import surname


def register_converter(data):
    if "_" in data:
        parts = data.split("_")
        return "".join(word.capitalize() for word in parts)
    else:
        string = ""
        for symbol in data:
            if symbol.isupper():
                string += "_" + symbol.lower()
            else:
                string += "".join(symbol)
        return string.strip("_")


print(register_converter(data = "otus_course"))
print(register_converter(data = "PythonIsTheBest"))
```

## Задание 2: Проверка валидности даты

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
def validate_date(date):
    if "." not in date and date.count(".") != 2:
        return False
    day, month, year = list(map(int, date.split(".")))

    if  (1 <= day <= 31) and (
            1 <= month <= 12) and 0 <= year:
        if is_leap_year(year):
            if month == 2 and day > 29:
                print("зашло")
                return False
        else:
            if month == 2 and day > 28:
                return False
        if day == 31 and month in (4,6,9,11):
            return False
        return True

    else:
        return False
def is_leap_year(year):
     return (year % 4 == 0) and (year % 100 != 0) or (year % 400 == 0)

print(validate_date("29.02.2000"))
print(validate_date("29.02.2001"))
print(validate_date("31.04.1962"))
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
def is_prime_number(number):
    if number < 2:
        return False
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True

print(is_prime_number(17))
print(is_prime_number(20))
print(is_prime_number(23))
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
def is_valid_name(name):
    return name.isalpha() and name.istitle()

def is_valid_age(age):
    return age.isdigit() and  18 <= int(age) <= 60

def is_valid_id(user_id):
    return  user_id.isdigit() and int(user_id) > 0 and user_id.zfill(8) not in ids

users = {}
ids = set()


while True:


    print(f"ids: {ids}")

    print("Введите данные пользователя: ")

    # Ввод имени
    name = input("Имя: ").capitalize()
    if not name:
        break
    if not is_valid_name(name):
        print("Ошибка: имя должно начинаться с заглавной буквы и содержать только буквы")
        continue

    # Ввод фамилии
    surname = input("Фамилия: ")
    if not surname:
        break
    if not is_valid_name(surname):
        print("Ошибка: фамилия должна начинаться с заглавной буквы и содержать только буквы")
        continue

    # Ввод возраста
    age = input("Возраст: ")
    if not is_valid_age(age):
        print("Ошибка: возраст должен быть целым числом в диапазоне от 18 до 60")
        continue

    # Ввод ID
    user_id = input("ID: ")
    print(user_id)
    if not is_valid_id(user_id):
        print("Ошибка: ID должен быть уникальным восьмизначным положительным числом")
        continue

    ids.add(user_id.zfill(8))

    users[user_id] = (name, surname, age)

print(users)
```
